Árboles de decisión
Los árboles de decisión son modelos de caja blanca, porque podemos ver y entender cómo llegan a una decisión. Funcionan como un árbol: empiezan en una raíz y van haciendo preguntas o evaluando condiciones hasta llegar a una respuesta.

Por ejemplo, si queremos saber si una persona va a comprar algo, el árbol puede preguntar primero: ¿Tiene más de 18 años? Dependiendo de la respuesta, tomamos una rama u otra. Después se pueden evaluar más condiciones hasta llegar a una predicción.

Entrenamiento
Para entrenar un árbol, el algoritmo busca las condiciones que mejor separan los datos. Dos métodos muy conocidos son CART y los métodos basados en entropía.

CART busca dividir los datos de la mejor manera posible, normalmente usando criterios como la impureza de Gini.

La entropía mide qué tan desordenados o mezclados están los datos. El árbol intenta reducir ese desorden al hacer las divisiones.

Predicciones y probabilidades
Cuando llega un dato nuevo, el árbol empieza desde la raíz, evalúa la condición y toma la rama que corresponde. Continúa así hasta llegar a una hoja, donde obtiene la predicción.

Además de predecir una clase, un árbol puede dar probabilidades. Por ejemplo, puede decir que hay un 80% de probabilidad de que un cliente compre y un 20% de que no compre.

Árboles de regresión
Los árboles también pueden utilizarse para regresión, cuando queremos predecir un número en lugar de una categoría. Por ejemplo, podemos usarlos para predecir el precio de una casa.

Restricciones y regularización
Un problema de los árboles es que pueden crecer demasiado y memorizar los datos de entrenamiento, lo que se conoce como sobreajuste.

Para evitarlo, podemos poner restricciones, como limitar la profundidad máxima del árbol o exigir un número mínimo de datos para hacer una división.

También podemos usar métodos de regularización, que ayudan a controlar la complejidad del modelo y evitar que el árbol sea demasiado grande.
