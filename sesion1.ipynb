AI 2 – Patrones complejos y no lineales

En esta parte entendí que la inteligencia artificial y el aprendizaje automático sirven para encontrar patrones en grandes cantidades de información.

Cuando hablamos de patrones complejos no lineales, significa que los datos no siempre tienen una relación sencilla. Por ejemplo, si tenemos información sobre estudiantes, no necesariamente significa que mientras más horas estudien, mejor será su nota de una manera perfectamente proporcional. Pueden existir muchos factores que influyan al mismo tiempo.

La IA puede analizar grandes volúmenes de datos y encontrar esas relaciones que serían muy difíciles de detectar manualmente.

Fuga de datos (Data Leakage)

La fuga de datos ocurre cuando información que no debería estar disponible para el modelo termina entrando en el entrenamiento.

Esto es un problema porque el modelo puede parecer muy bueno durante las pruebas, pero cuando recibe datos realmente nuevos, funciona mal.

Por ejemplo, imaginemos que queremos predecir si un estudiante va a aprobar. Si durante el entrenamiento le damos accidentalmente información que solamente se conoce después de que el estudiante presenta el examen, el modelo estaría haciendo "trampa".

Por eso entendí que debemos separar correctamente los datos y asegurarnos de que el modelo solamente utilice la información que realmente tendría disponible en el momento de hacer una predicción.

Idea fácil:

Fuga de datos = el modelo recibe información que no debería conocer.

Imputación

La imputación se utiliza cuando tenemos datos faltantes.

Por ejemplo:

Edad	Nota
20	4.5
21	3.8
—	4.0
23	4.7

Aquí falta una edad.

En lugar de simplemente eliminar ese registro, podemos reemplazar el dato faltante utilizando algún método. Por ejemplo, podemos utilizar la media, la mediana u otros métodos más avanzados.

La idea es preparar los datos para que el algoritmo pueda trabajar con ellos.

Idea fácil:

Imputar = completar los datos que faltan de una manera razonable.

Escalamiento

El escalamiento sirve para poner las variables en escalas comparables.

Por ejemplo, podemos tener:

Edad: 18–60
Salario: 1.000–10.000.000
Número de hijos: 0–5

Las variables tienen valores muy diferentes.

Algunos algoritmos pueden verse afectados por estas diferencias, por lo que podemos transformar los datos para que tengan escalas similares.

Un ejemplo sería llevar los valores a un rango como 0 a 1.

Idea fácil:

Escalar = hacer que las variables estén en una escala parecida para que el modelo pueda trabajar mejor.

Compresión o reducción de dimensiones

Cuando tenemos muchísimas variables, el modelo puede volverse más complicado.

Por ejemplo, podemos tener una base de datos con 500 características. Muchas de ellas pueden aportar poca información o incluso ser repetitivas.

La reducción de dimensiones busca quedarse con la información más importante utilizando menos variables.

Esto puede ayudar a:

Reducir el tamaño de los datos.
Hacer que el modelo sea más rápido.
Facilitar la visualización.
Evitar cierta complejidad innecesaria.

Idea fácil:

Reducir dimensiones = quedarnos con lo importante sin cargar al modelo con información innecesaria.

Clasificador

Un clasificador es un modelo que intenta colocar cada dato en una categoría.

Por ejemplo:

Correo → Spam / No spam
Paciente → Riesgo alto / Riesgo bajo
Imagen → Perro / Gato
Estudiante → Aprueba / No aprueba

El modelo aprende utilizando ejemplos anteriores y después intenta clasificar datos nuevos.

Idea fácil:

Clasificar = decidir a qué categoría pertenece un dato.

El truco del Kernel

El truco del kernel se utiliza principalmente en modelos como las máquinas de vectores de soporte (SVM).

La idea que entendí es que algunos datos no se pueden separar fácilmente en su espacio original.

Imaginemos puntos de dos clases formando círculos:

🔵 🔵 🔵
 🔴 🔴
🔵 🔴 🔵

Una línea recta posiblemente no sea suficiente para separar las clases.

El kernel permite trabajar matemáticamente como si los datos fueran llevados a un espacio diferente, donde sea más fácil encontrar una separación.

Lo interesante es que el algoritmo puede hacer esto sin tener que calcular directamente todas las nuevas dimensiones.

Por eso se llama "truco del kernel".

Idea fácil:

Kernel = una forma inteligente de transformar el problema para poder separar mejor los datos.

Bagging

Bagging significa Bootstrap Aggregating.

La idea principal es utilizar varios modelos en lugar de depender de uno solo.

Por ejemplo, podemos crear varios árboles de decisión utilizando diferentes muestras de los datos:

Árbol 1
Árbol 2
Árbol 3
Árbol 4
Árbol 5

Después combinamos sus resultados.

Esto ayuda a que el resultado final sea más estable y reduce el riesgo de depender demasiado de un solo árbol.

Un ejemplo muy conocido es Random Forest, que utiliza muchos árboles de decisión.

Idea fácil:

Bagging = varios modelos trabajan juntos para obtener una decisión más estable.

Sesgo y varianza

Estos dos conceptos son importantes para entender por qué un modelo puede funcionar bien o mal.

Sesgo (Bias)

El sesgo ocurre cuando el modelo es demasiado simple y no logra aprender correctamente los patrones de los datos.

Por ejemplo, si intentamos representar una relación muy complicada utilizando un modelo demasiado sencillo.

Esto puede producir underfitting.

Idea fácil:

Mucho sesgo = el modelo es demasiado simple y no aprende lo suficiente.

Varianza

La varianza ocurre cuando el modelo se adapta demasiado a los datos de entrenamiento.

Es como si el modelo se aprendiera los ejemplos de memoria, incluyendo sus errores o particularidades.

Cuando recibe datos nuevos, puede funcionar peor.

Esto se relaciona con el overfitting.

GridSearchCV

GridSearchCV sirve para buscar los mejores parámetros de un modelo.

Los modelos de machine learning tienen parámetros que podemos configurar. Por ejemplo, en un árbol podemos cambiar su profundidad.

GridSearchCV prueba diferentes combinaciones.

Por ejemplo:

profundidad = 3
profundidad = 5
profundidad = 10

Y también puede probar otros parámetros al mismo tiempo.

El problema es que si tenemos muchas posibilidades, tiene que probar muchísimas combinaciones, por lo que puede ser costoso en tiempo y recursos.

Idea fácil:

GridSearchCV = probar todas las combinaciones posibles de parámetros.

RandomizedSearchCV

RandomizedSearchCV hace algo parecido, pero en lugar de probar absolutamente todas las combinaciones, selecciona combinaciones al azar dentro de las posibilidades.

Por eso puede ser mucho más rápido cuando tenemos muchos parámetros.

No necesariamente encuentra la combinación perfecta, pero puede encontrar una muy buena combinación gastando menos tiempo y recursos.

Entonces:

GridSearchCV: prueba sistemáticamente todas las combinaciones.
RandomizedSearchCV: prueba solamente algunas combinaciones seleccionadas.

Idea fácil:

RandomizedSearchCV = buscar una buena configuración sin tener que probar absolutamente todo.

Diseño arquitectónico con Python

El diseño arquitectónico lo entendí como la forma en que organizamos todo nuestro proyecto de inteligencia artificial.

No se trata solamente de escribir código en Python. También debemos pensar cómo van a funcionar las diferentes partes del sistema.

Por ejemplo:

Datos → Limpieza → Preparación → Modelo → Evaluación → Predicción

La ventaja de organizarlo así es que el proyecto se vuelve más fácil de entender, mantener y modificar.
