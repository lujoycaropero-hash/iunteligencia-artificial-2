Inicialización y funciones de activación

La inicialización es el proceso de poner valores iniciales a los pesos de una red neuronal antes de empezar el entrenamiento. Es importante porque una mala inicialización puede hacer que los gradientes se vuelvan muy pequeños o demasiado grandes mientras se van propagando por las diferentes capas.

Cuando el gradiente se hace muy pequeño, se produce el desvanecimiento del gradiente, por lo que las primeras capas aprenden muy lentamente. En cambio, si el gradiente aumenta demasiado, puede aparecer el gradiente explosivo, haciendo que el entrenamiento sea inestable.

Por eso, una de las ideas principales de la inicialización es mantener las varianzas de las entradas y salidas más o menos estables entre las capas. Así se evita que las señales se hagan demasiado pequeñas o demasiado grandes. Para esto existen métodos como Xavier/Glorot, que se usa bastante con funciones como tanh, y He, que está pensado principalmente para funciones como ReLU.

También han aparecido diferentes variantes de las funciones de activación para solucionar algunos problemas de las funciones más antiguas. Algunas funciones utilizadas en arquitecturas actuales son ReLU, ELU, SELU, GELU, Swish y Mish. Entre ellas hay funciones que no son monótonas, es decir, que no siempre aumentan cuando aumenta la entrada.

Estas funciones permiten que la red tenga un comportamiento más flexible y pueden ayudar al entrenamiento. Por ejemplo, GELU y Swish tienen una transición más suave que ReLU y se utilizan en varias arquitecturas modernas.

De todas formas, la estabilidad de una red no depende solamente de la función de activación. También influyen la inicialización de los pesos, la normalización, las conexiones residuales y el diseño general de la arquitectura. Todo esto ayuda a que los gradientes puedan propagarse correctamente sin desaparecer o crecer demasiado.

En resumen, una buena inicialización y una función de activación adecuada ayudan a mantener estable el entrenamiento y permiten que la información y los gradientes se propaguen correctamente por toda la red.
